```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

print(f"TensorFlow Version: {tf.__version__}")
```

```python
# Load all datasets
train_df = pd.read_csv("data/train_FD001.txt", sep=r"\s+", header=None, names=cols)
test_df = pd.read_csv("data/test_FD001.txt", sep=r"\s+", header=None, names=cols)
rul_df = pd.read_csv("data/RUL_FD001.txt", sep=r"\s+", header=None, names=["RUL"])

print(f"Train shape: {train_df.shape}")
print(f"Test shape: {test_df.shape}")
print(f"RUL shape: {rul_df.shape}")
print(f"\nNumber of engines in train: {train_df['engine_id'].nunique()}")
print(f"Number of engines in test: {test_df['engine_id'].nunique()}")
```

```python
# Calculate RUL for training data
# RUL = max_cycle - current_cycle for each engine
train_df['RUL'] = train_df.groupby('engine_id')['cycle'].transform('max') - train_df['cycle']

# Clip RUL to maximum of 125 (common practice in CMAPSS literature)
train_df['RUL'] = train_df['RUL'].clip(upper=125)

print("RUL Statistics:")
print(train_df['RUL'].describe())
print(f"\nSample data with RUL:")
print(train_df[['engine_id', 'cycle', 'RUL']].head(10))
```

```python
# Select sensor features (remove constant or near-constant sensors)
# Remove engine_id, cycle, and select informative sensors
sensor_cols = [f"sensor{i}" for i in range(1, 22)]
setting_cols = ["setting1", "setting2", "setting3"]

# Check variance to identify useful features
feature_variance = train_df[sensor_cols + setting_cols].var()
useful_features = feature_variance[feature_variance > 0.01].index.tolist()

print(f"Selected features ({len(useful_features)}): {useful_features}")

# Add operational settings to features
feature_cols = useful_features
```

```python
# Normalize features using MinMaxScaler
scaler = MinMaxScaler()

# Fit on training data
train_df[feature_cols] = scaler.fit_transform(train_df[feature_cols])

# Transform test data using same scaler
test_df[feature_cols] = scaler.transform(test_df[feature_cols])

print("Normalization complete!")
print(f"Feature range: {train_df[feature_cols].min().min():.4f} to {train_df[feature_cols].max().max():.4f}")
```

```python
# Create sequences for LSTM
# Use sliding window approach: use past 'sequence_length' cycles to predict RUL

def create_sequences(df, sequence_length, feature_cols, target_col='RUL'):
    """
    Create sequences from time series data for LSTM
    """
    sequences = []
    targets = []
    
    for engine_id in df['engine_id'].unique():
        engine_data = df[df['engine_id'] == engine_id]
        
        # Extract features and target
        features = engine_data[feature_cols].values
        target = engine_data[target_col].values
        
        # Create sequences
        for i in range(len(features) - sequence_length + 1):
            sequences.append(features[i:i+sequence_length])
            targets.append(target[i+sequence_length-1])  # Predict RUL at end of sequence
    
    return np.array(sequences), np.array(targets)

# Set sequence length (number of time steps to look back)
SEQUENCE_LENGTH = 30

print(f"Creating sequences with length {SEQUENCE_LENGTH}...")
X_train, y_train = create_sequences(train_df, SEQUENCE_LENGTH, feature_cols, 'RUL')

print(f"X_train shape: {X_train.shape}")  # (samples, timesteps, features)
print(f"y_train shape: {y_train.shape}")
```

```python
# Prepare test sequences
# For test data, we need the last 'sequence_length' cycles of each engine

def create_test_sequences(df, sequence_length, feature_cols):
    """
    Create test sequences - use last 'sequence_length' cycles for each engine
    """
    sequences = []
    
    for engine_id in df['engine_id'].unique():
        engine_data = df[df['engine_id'] == engine_id]
        features = engine_data[feature_cols].values
        
        # Take last sequence_length cycles (or all if less than sequence_length)
        if len(features) >= sequence_length:
            sequences.append(features[-sequence_length:])
        else:
            # Pad with zeros if not enough data
            padding = np.zeros((sequence_length - len(features), len(feature_cols)))
            sequences.append(np.vstack([padding, features]))
    
    return np.array(sequences)

X_test = create_test_sequences(test_df, SEQUENCE_LENGTH, feature_cols)
y_test = rul_df['RUL'].values

print(f"X_test shape: {X_test.shape}")
print(f"y_test shape: {y_test.shape}")
```

```python
# Build LSTM Model
def build_lstm_model(input_shape):
    """
    Build a deep LSTM model for RUL prediction
    """
    model = Sequential([
        # First LSTM layer with return sequences
        LSTM(128, return_sequences=True, input_shape=input_shape),
        Dropout(0.2),
        
        # Second LSTM layer with return sequences
        LSTM(64, return_sequences=True),
        Dropout(0.2),
        
        # Third LSTM layer
        LSTM(32, return_sequences=False),
        Dropout(0.2),
        
        # Dense layers
        Dense(32, activation='relu'),
        Dropout(0.2),
        
        Dense(16, activation='relu'),
        
        # Output layer - predicting single RUL value
        Dense(1)
    ])
    
    # Compile model
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=0.001),
        loss='mse',
        metrics=['mae']
    )
    
    return model

# Create model
input_shape = (X_train.shape[1], X_train.shape[2])  # (timesteps, features)
model = build_lstm_model(input_shape)

# Display model architecture
model.summary()
```

```python
# Setup callbacks for training
early_stopping = EarlyStopping(
    monitor='val_loss',
    patience=15,
    restore_best_weights=True,
    verbose=1
)

reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=5,
    min_lr=0.00001,
    verbose=1
)

callbacks = [early_stopping, reduce_lr]

print("Callbacks configured:")
```

```python
# Train the model
print("Starting model training...")
print("=" * 60)

history = model.fit(
    X_train, y_train,
    epochs=100,
    batch_size=64,
    validation_split=0.2,
    callbacks=callbacks,
    verbose=1
)

print("\n" + "=" * 60)
print("Training completed!")
```

```python
# Plot training history
plt.figure(figsize=(14, 5))

# Loss plot
plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE)')
plt.title('Model Loss During Training')
plt.legend()
plt.grid(True, alpha=0.3)

# MAE plot
plt.subplot(1, 2, 2)
plt.plot(history.history['mae'], label='Training MAE')
plt.plot(history.history['val_mae'], label='Validation MAE')
plt.xlabel('Epoch')
plt.ylabel('Mean Absolute Error')
plt.title('Model MAE During Training')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Make predictions on test set
y_pred = model.predict(X_test)
y_pred = y_pred.flatten()  # Convert to 1D array

# Calculate metrics
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("=" * 60)
print("MODEL PERFORMANCE ON TEST SET")
print("=" * 60)
print(f"Root Mean Squared Error (RMSE): {rmse:.2f} cycles")
print(f"Mean Absolute Error (MAE):      {mae:.2f} cycles")
print(f"R² Score:                        {r2:.4f}")
print("=" * 60)
```

```python
# Visualize predictions vs actual RUL
plt.figure(figsize=(15, 5))

# Plot 1: Scatter plot
plt.subplot(1, 2, 1)
plt.scatter(y_test, y_pred, alpha=0.6, edgecolors='k', linewidth=0.5)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 
         'r--', lw=2, label='Perfect Prediction')
plt.xlabel('Actual RUL (cycles)')
plt.ylabel('Predicted RUL (cycles)')
plt.title('Predicted vs Actual RUL')
plt.legend()
plt.grid(True, alpha=0.3)

# Plot 2: Error distribution
plt.subplot(1, 2, 2)
errors = y_pred - y_test
plt.hist(errors, bins=30, edgecolor='black', alpha=0.7)
plt.xlabel('Prediction Error (cycles)')
plt.ylabel('Frequency')
plt.title(f'Prediction Error Distribution\nMean Error: {np.mean(errors):.2f} cycles')
plt.axvline(x=0, color='r', linestyle='--', linewidth=2, label='Zero Error')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Show sample predictions
results_df = pd.DataFrame({
    'Engine_ID': range(1, len(y_test) + 1),
    'Actual_RUL': y_test,
    'Predicted_RUL': y_pred.round(2),
    'Error': (y_pred - y_test).round(2)
})

print("\nSample Predictions (First 10 Engines):")
print("=" * 70)
print(results_df.head(10).to_string(index=False))
print("\n" + "=" * 70)
print("\nSample Predictions (Engines with largest errors):")
print("=" * 70)
print(results_df.nlargest(5, 'Error', keep='all')[['Engine_ID', 'Actual_RUL', 'Predicted_RUL', 'Error']].to_string(index=False))
```

```python
# 7. Model Performance Visualization (if model predictions available)
try:
    # Check if predictions exist
    if 'y_pred' in locals() or 'y_pred_loaded' in locals():
        y_predictions = y_pred_loaded if 'y_pred_loaded' in locals() else y_pred
        
        fig = plt.figure(figsize=(16, 10))
        gs = GridSpec(2, 3, figure=fig, hspace=0.3, wspace=0.3)
        
        # 1. Prediction vs Actual Scatter with Density
        ax1 = fig.add_subplot(gs[0, :2])
        scatter = ax1.scatter(y_test, y_predictions, c=np.abs(y_test - y_predictions), 
                             cmap='RdYlGn_r', alpha=0.6, edgecolors='black', linewidth=0.5, s=60)
        ax1.plot([0, max(y_test.max(), y_predictions.max())], 
                [0, max(y_test.max(), y_predictions.max())], 
                'r--', lw=3, label='Perfect Prediction', alpha=0.8)
        ax1.set_xlabel('Actual RUL (cycles)', fontsize=12, fontweight='bold')
        ax1.set_ylabel('Predicted RUL (cycles)', fontsize=12, fontweight='bold')
        ax1.set_title('Model Predictions vs Actual RUL', fontsize=14, fontweight='bold')
        ax1.legend(fontsize=11, loc='upper left')
        ax1.grid(True, alpha=0.3)
        cbar = plt.colorbar(scatter, ax=ax1)
        cbar.set_label('Absolute Error (cycles)', fontsize=10, fontweight='bold')
        
        # 2. Error Distribution
        ax2 = fig.add_subplot(gs[0, 2])
        errors = y_predictions - y_test
        ax2.hist(errors, bins=40, edgecolor='black', alpha=0.75, color='steelblue')
        ax2.axvline(x=0, color='red', linestyle='--', linewidth=2.5, label='Zero Error')
        ax2.axvline(x=errors.mean(), color='green', linestyle='--', linewidth=2, 
                   label=f'Mean: {errors.mean():.2f}')
        ax2.set_xlabel('Prediction Error (cycles)', fontsize=11, fontweight='bold')
        ax2.set_ylabel('Frequency', fontsize=11, fontweight='bold')
        ax2.set_title('Error Distribution', fontsize=12, fontweight='bold')
        ax2.legend(fontsize=9)
        ax2.grid(True, alpha=0.3, axis='y')
        
        # 3. Accuracy by RUL Range
        ax3 = fig.add_subplot(gs[1, 0])
        rul_bins = [0, 30, 60, 90, 125]
        rul_labels = ['0-30', '31-60', '61-90', '91-125']
        errors_by_rul = []
        
        for i in range(len(rul_bins)-1):
            mask = (y_test >= rul_bins[i]) & (y_test < rul_bins[i+1])
            if mask.sum() > 0:
                errors_by_rul.append(np.abs(errors[mask]))
            else:
                errors_by_rul.append([])
        
        bp = ax3.boxplot(errors_by_rul, labels=rul_labels, patch_artist=True,
                        boxprops=dict(facecolor='lightcoral', alpha=0.7, linewidth=1.5),
                        medianprops=dict(color='darkred', linewidth=2),
                        whiskerprops=dict(linewidth=1.5))
        ax3.set_xlabel('RUL Range (cycles)', fontsize=11, fontweight='bold')
        ax3.set_ylabel('Absolute Error (cycles)', fontsize=11, fontweight='bold')
        ax3.set_title('Prediction Error by RUL Range', fontsize=12, fontweight='bold')
        ax3.grid(True, alpha=0.3, axis='y')
        
        # 4. Cumulative Accuracy
        ax4 = fig.add_subplot(gs[1, 1])
        abs_errors = np.abs(errors)
        thresholds = np.arange(0, 50, 2)
        accuracies = [np.sum(abs_errors <= t) / len(abs_errors) * 100 for t in thresholds]
        
        ax4.plot(thresholds, accuracies, linewidth=3, color='green', marker='o', 
                markersize=5, alpha=0.8)
        ax4.fill_between(thresholds, accuracies, alpha=0.3, color='green')
        ax4.set_xlabel('Error Tolerance (cycles)', fontsize=11, fontweight='bold')
        ax4.set_ylabel('Cumulative Accuracy (%)', fontsize=11, fontweight='bold')
        ax4.set_title('Cumulative Accuracy Curve', fontsize=12, fontweight='bold')
        ax4.grid(True, alpha=0.3)
        ax4.axhline(y=50, color='red', linestyle='--', alpha=0.5, label='50% accuracy')
        ax4.axhline(y=80, color='orange', linestyle='--', alpha=0.5, label='80% accuracy')
        ax4.legend(fontsize=9)
        
        # 5. Performance Metrics Summary
        ax5 = fig.add_subplot(gs[1, 2])
        ax5.axis('off')
        
        # Calculate metrics
        rmse_val = np.sqrt(mean_squared_error(y_test, y_predictions))
        mae_val = mean_absolute_error(y_test, y_predictions)
        r2_val = r2_score(y_test, y_predictions)
        within_10 = np.sum(abs_errors <= 10) / len(abs_errors) * 100
        within_20 = np.sum(abs_errors <= 20) / len(abs_errors) * 100
        
        metrics_text = f"""
        Performance Metrics
        
        RMSE: {rmse_val:.2f} cycles
        MAE: {mae_val:.2f} cycles
        R² Score: {r2_val:.4f}
        
        Accuracy:
        ±10 cycles: {within_10:.1f}%
        ±20 cycles: {within_20:.1f}%
        
        Max Error: {abs_errors.max():.2f}
        Min Error: {abs_errors.min():.2f}
        Std Dev: {abs_errors.std():.2f}
        """
        
        ax5.text(0.5, 0.5, metrics_text, fontsize=11, ha='center', va='center',
                bbox=dict(boxstyle='round,pad=1.5', facecolor='lightyellow', 
                         alpha=0.9, edgecolor='black', linewidth=2),
                family='monospace', fontweight='bold')
        
        plt.suptitle('NASA CMAPSS Model: Comprehensive Performance Analysis', 
                    fontsize=18, fontweight='bold', y=0.98)
        plt.savefig('cmapss_model_performance.png', dpi=300, bbox_inches='tight')
        plt.show()
        
        print("✓ Figure 7: Model Performance Analysis saved as 'cmapss_model_performance.png'")
    else:
        print("⚠ No model predictions found. Train the model first to generate performance visualizations.")
        
except Exception as e:
    print(f"⚠ Could not generate model performance visualization: {e}")
    print("This is normal if the model hasn't been trained yet.")
```
